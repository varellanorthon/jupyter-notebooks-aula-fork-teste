# Funções e Interfaces

Este capítulo apresenta um módulo chamado `jupyturtle`, que permite criar desenhos simples dando instruções a uma tartaruga imaginária.
Vamos usar esse módulo para escrever funções que desenham quadrados, polígonos e círculos -- e para demonstrar **design de interface**, que é uma forma de projetar funções que trabalham juntas.

## O módulo jupyturtle

Para usar o módulo `jupyturtle`, podemos importá-lo assim.

```python
import jupyturtle
```

Agora podemos usar as funções definidas no módulo, como `make_turtle` e `forward`.

`make_turtle` cria um **canvas**, que é um espaço na tela onde podemos desenhar, e uma tartaruga, representada por uma carapaça circular e uma cabeça triangular.
O círculo mostra a posição da tartaruga e o triângulo indica a direção para a qual ela está voltada.

`forward` move a tartaruga uma determinada distância na direção em que ela está voltada, desenhando um segmento de reta pelo caminho.
A distância é medida em unidades arbitrárias -- o tamanho real depende da tela do seu computador.

Vamos usar funções definidas no módulo `jupyturtle` muitas vezes, então seria bom não precisar escrever o nome do módulo toda vez.
Isso é possível se importarmos o módulo desta forma.

```python
from jupyturtle import make_turtle, forward
```

Essa versão do comando de importação importa `make_turtle` e `forward` do módulo `jupyturtle`, de modo que podemos chamá-las assim.

```python
make_turtle()
forward(100)
```

`jupyturtle` fornece outras duas funções que vamos usar, chamadas `left` e `right`.
Vamos importá-las assim.

```python
from jupyturtle import left, right
```

`left` faz a tartaruga virar para a esquerda. Ela recebe um argumento, que é o ângulo do giro em graus.
Por exemplo, podemos fazer um giro de 90 graus para a esquerda assim.

```python
make_turtle()
forward(50)
left(90)

forward(50)
```

Esse programa move a tartaruga para o leste e depois para o norte, deixando dois segmentos de reta para trás.
Antes de continuar, veja se você consegue modificar o programa anterior para fazer um quadrado.

## Fazendo um quadrado

Aqui está uma forma de fazer um quadrado.

```python
make_turtle()

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)

forward(50)
left(90)
```

Como esse programa repete o mesmo par de linhas quatro vezes, podemos fazer a mesma coisa de forma mais concisa com um laço `for`.

```python
make_turtle()

for i in range(4):
  forward(50)
  left(90)
```

## Encapsulamento e generalização

Vamos pegar o código de desenhar o quadrado da seção anterior e colocá-lo em uma função chamada `square`.

```python
def quadrado():
  for i in range(4):
    forward(50)
    left(90)
```

Agora podemos chamar a função assim.

```python
make_turtle()
quadrado()
```

Colocar um trecho de código dentro de uma função é chamado de **encapsulamento**.
Um dos benefícios do encapsulamento é que ele **atribui um nome ao código**, o que serve como uma espécie de documentação. Outra vantagem é que, se você reutilizar o código, é mais conciso chamar a função duas vezes do que copiar e colar o corpo dela!

Na versão atual, o tamanho do quadrado é sempre `50`.
Se quisermos desenhar quadrados de tamanhos diferentes, podemos receber o comprimento dos lados como parâmetro.

```python
def quadrado(tam: int):
  for i in range(4):
    forward(tam)
    left(90)
```

Agora podemos desenhar quadrados de tamanhos diferentes.

```python
make_turtle()

quadrado(40)
quadrado(30)
```

Adicionar um parâmetro a uma função é chamado de **generalização**, porque torna a função mais geral: na versão anterior, o quadrado tinha sempre o mesmo tamanho; nesta versão ele pode ter qualquer tamanho.

Se adicionarmos outro parâmetro, podemos torná-la ainda mais geral.
A função a seguir desenha polígonos regulares com um determinado número de lados.

```python
def polygon(n, length):
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
```

Em um polígono regular com `n` lados, o ângulo entre lados adjacentes é `360 / n` graus.

O exemplo a seguir desenha um polígono de `7` lados com comprimento de lado `30`.

```python
make_turtle()
polygon(7, 30)
```

Quando uma função tem mais do que alguns argumentos numéricos, é fácil esquecer quais são eles, ou em que ordem devem aparecer.
Pode ser uma boa ideia incluir os nomes dos parâmetros na lista de argumentos.

```python
make_turtle()
polygon(n=7, length=30)
```

Esses são às vezes chamados de "argumentos nomeados", porque incluem os nomes dos parâmetros.
Mas em Python eles são mais frequentemente chamados de **argumentos por palavra-chave** (keyword arguments) (para não confundir com palavras-chave do Python, como `for` e `def`).

Esse uso do operador de atribuição, `=`, é um lembrete de como argumentos e parâmetros funcionam: quando você chama uma função, os argumentos são atribuídos aos parâmetros.

## Um plano de desenvolvimento

Um **plano de desenvolvimento** é um processo para escrever programas.
O processo que usamos neste capítulo é "encapsulamento e generalização".
As etapas desse processo são:

1.  Comece escrevendo um programa pequeno, sem definições de função.

2.  Uma vez que o programa funcione, identifique uma parte coerente dele, encapsule essa parte em uma função e dê um nome a ela.

3.  Generalize a função adicionando parâmetros apropriados.

4.  Repita os passos 1 a 3 até ter um conjunto de funções funcionando.

5.  Procure oportunidades de melhorar o programa por meio de refatoração. Por exemplo, se você tiver código semelhante em vários lugares, considere transformá-lo em uma função apropriadamente geral.

Esse processo tem algumas desvantagens mas pode ser útil se você não souber de antemão como dividir o programa em funções. Essa abordagem permite que você projete o programa aos poucos, conforme avança.

O design de uma função tem duas partes:

* A **interface** é como a função é usada, incluindo seu nome, os parâmetros que ela recebe e o que ela deve fazer.

* A **implementação** é como a função faz o que deve fazer.

## Docstrings

Uma **docstring** é uma string no início de uma função que explica a interface ("doc" é abreviação de "documentação").
Aqui está um exemplo:

```python
def polygon(n, length):
    """Desenha segmentos de linha com o comprimento informado.

    n: número inteiro de segmentos de linha
    length: comprimento dos segmentos de linha
    """
    angle = 360 / n
    for i in range(n):
        forward(length)
        left(angle)
```

Por convenção, docstrings são strings delimitadas por aspas triplas, também conhecidas como **strings multilinha**, porque as aspas triplas permitem que a string ocupe mais de uma linha.

Uma docstring deve:

* Explicar de forma concisa **o que a função faz**, sem entrar em detalhes de como ela funciona,

* Explicar que efeito **cada parâmetro tem sobre o comportamento** da função, e

* Indicar **qual tipo** cada parâmetro deve ter, se isso não for óbvio.

Escrever esse tipo de documentação é uma parte importante do design de interfaces.
Uma interface bem projetada deve ser simples de explicar; se você tiver dificuldade para explicar uma de suas funções, talvez a interface possa ser melhorada.

## Operadores lógicos

Para combinar valores booleanos em expressões, podemos usar **operadores lógicos**.
Os mais comuns são `and`, `or` e `not`.
O significado desses operadores é parecido com o seu significado em português.
Por exemplo, o valor da expressão a seguir é `True` somente se `x` for maior que `0` *e* menor que `10`.

```python
x = 5
y = 7
```

```python
x > 0 and x < 10 # x > 0 && x <10
```

A expressão a seguir é `True` se *uma ou ambas* as condições forem verdadeiras, ou seja, se o número é divisível por 2 *ou* por 3:

```python
x % 2 == 0 or x % 3 == 0 # x % 2 == 0 || x % 3 == 0
```

Por fim, o operador `not` nega uma expressão booleana, então a expressão a seguir é `True` se `x > y` for `False`.

```python
not x > y
```

Estritamente falando, os operandos de um operador lógico deveriam ser expressões booleanas, mas o Python não é muito rígido quanto a isso.
Qualquer número diferente de zero é interpretado como `True`:

```python
9090 and True
```

Essa flexibilidade pode ser útil, mas há algumas sutilezas que podem gerar confusão.
Talvez seja melhor evitá-la.

## Comandos if

Para escrever programas úteis, quase sempre precisamos da capacidade de
verificar condições e mudar o comportamento do programa de acordo.
**Comandos condicionais** nos dão essa capacidade. A forma mais simples é
o comando `if`:

```python
if x > 0:
    print('x is positive')
```

`if` é uma palavra-chave do Python.
Comandos `if` têm a mesma estrutura que definições de função: um
cabeçalho seguido de um comando indentado, ou sequência de comandos, chamada de **bloco**.

A expressão booleana depois do `if` é chamada de **condição**.
Se ela for verdadeira, os comandos do bloco indentado são executados. Se não, eles não são.

Não há limite para o número de comandos que podem aparecer no bloco, mas precisa haver pelo menos um.
De vez em quando, é útil ter um bloco que não faz nada, geralmente como um marcador de lugar para um código que você ainda não escreveu.
Nesse caso, você pode usar o comando `pass`, que não faz nada.

```python
if x < 0:
    pass          # TODO: ajustar para valores negativos
```

A palavra `TODO` em um comentário é um lembrete convencional de que há algo que você precisa fazer depois.

## A cláusula `else`

Um comando `if` pode ter uma segunda parte, chamada de cláusula `else`.
A sintaxe é assim:

```python
if x % 2 == 0:
    print('x is even')
else:
    print('x is odd')
```

Se a condição for verdadeira, o primeiro comando indentado é executado; caso contrário, o segundo comando indentado é executado.

Neste exemplo, se `x` for par, o resto da divisão de `x` por `2` é `0`, então a condição é verdadeira e o programa exibe `x is even`.
Se `x` for ímpar, o resto é `1`, então a condição
é falsa, e o programa exibe `x is odd`.

Como a condição precisa ser verdadeira ou falsa, exatamente uma das alternativas será executada.
As alternativas são chamadas de **ramos** (branches).

## Condicionais encadeadas

Às vezes há mais de duas possibilidades e precisamos de mais de dois ramos.
Uma forma de expressar um cálculo assim é uma **condicional encadeada**, que inclui uma cláusula `elif`.

```python
if x < y:
    print('x é menor que y')
elif x > y:
    print('x é maior que y')
else:
    print('x e y são iguais')
```

`elif` é uma abreviação de "else if" ("senão se").
Não há limite para o número de cláusulas `elif`.
Se houver uma cláusula `else`, ela precisa estar no final, mas não precisa
haver uma.

Cada condição é verificada em ordem.
Se a primeira for falsa, a próxima é verificada, e assim por diante.
Se uma delas for verdadeira, o ramo correspondente é executado e o comando `if` termina.
Mesmo que mais de uma condição seja verdadeira, apenas o primeiro ramo verdadeiro é executado.

## Condicionais aninhadas

Um condicional também pode estar aninhado dentro de outro.
Poderíamos ter escrito o exemplo da seção anterior assim:

```python
if x == y:
    print('x e y são iguais')
else:
    if x < y:
        print('x é menor que y')
    else:
        print('x é maior que y')
```

O comando `if` externo contém dois ramos.
O primeiro ramo contém um comando simples. O segundo ramo contém outro comando `if`, que tem seus próprios dois ramos.
Esses dois ramos são ambos comandos simples, embora também pudessem ser comandos condicionais.

Embora a indentação dos comandos deixe a estrutura aparente, **condicionais aninhados** podem ser difíceis de ler.
Sugiro que você os evite quando puder.

Operadores lógicos costumam oferecer uma forma de simplificar comandos condicionais aninhados.
Aqui está um exemplo com um condicional aninhado.

```python
if 0 < x:
    if x < 10:
        print('x é um número positivo com apenas um dígito.')
```

O comando `print` só é executado se passarmos pelos dois condicionais, então obtemos o mesmo efeito com o operador `and`.

```python
if 0 < x and x < 10:
    print('x é um número positivo com apenas um dígito.')
```

Para esse tipo de condição, o Python oferece uma opção mais concisa:

```python
if 0 < x < 10:
    print('x é um número positivo com apenas um dígito.')
```

## Glossário

**expressão booleana:**
Uma expressão cujo valor é `True` ou `False`.

**operador lógico:**
Um dos operadores que combinam expressões booleanas, incluindo `and`, `or` e `not`.

**comando condicional:**
Um comando que controla o fluxo de execução dependendo de alguma condição.

**condição:**
A expressão booleana em um comando condicional que determina qual ramo é executado.

**bloco:**
Um ou mais comandos indentados para indicar que fazem parte de outro comando.

**ramo:**
Uma das sequências alternativas de comandos em um comando condicional.

**condicional encadeada:**
Um comando condicional com uma série de ramos alternativos.

**condicional aninhada:**
Um comando condicional que aparece em um dos ramos de outro comando condicional.

# Valores de Retorno

Quando você **chama uma função**, ela pode **retornar um valor** que você pode atribuir a uma variável ou usar como parte de uma expressão.

## Algumas funções têm valores de retorno

Quando você chama uma função como `math.sqrt`, o resultado é chamado de **valor de retorno**. Se a chamada da função aparecer no final de uma célula, o Jupyter exibe o valor de retorno imediatamente.

```python
import math

math.sqrt(42 / math.pi)
```

Se você atribuir o valor de retorno a uma variável, ele não é exibido.

```python
radius = math.sqrt(42 / math.pi)
```

Mas você pode exibi-lo depois.

```python
radius
```

Ou você pode usar o valor de retorno como parte de uma expressão.

```python
radius + math.sqrt(42 / math.pi)
```

Aqui está um exemplo de uma função que retorna um valor.

```python
def area_circulo(radius):
  area = math.pi * radius**2
  return area
```

`area_circulo` recebe `radius` como parâmetro e calcula a área de um círculo com esse raio.

A última linha é um comando `return` que retorna o valor de `area`.

Se chamarmos a função assim, o Jupyter exibe o valor de retorno.


```python
area_circulo(radius)
```

Podemos atribuir o valor de retorno a uma variável.

```python
area = area_circulo(radius)
```

Ou usá-lo como parte de uma expressão.

```python
area_circulo(radius) + 2 * area_circulo(radius / 2)
```

Mais tarde podemos exibir o valor da variável à qual atribuímos o resultado.

```python
area
```

## E algumas retornam None

Se uma função não tiver um comando `return`, ela retorna `None`, que é um valor especial como `True` e `False`.
Por exemplo, a função `repeat`.

```python
def repeat(word, n):
    print(word * n)
```

Se a chamarmos assim, ela exibe a palavra repetidas vezes, com base no parâmetro `n`

```python
repeat('Hello, ', 3)
```

Essa função usa a função `print` para exibir uma string, mas não usa um comando `return` para retornar um valor.
Se atribuirmos o resultado a uma variável, ela exibe a string mesmo assim.

```python
result = repeat('Hello, ', 3)
```

E se exibirmos o valor da variável, não obtemos nada.

```python
result
```

`result` de fato tem um valor, mas o Jupyter não o mostra.
No entanto, podemos exibi-lo assim.

```python
print(result)
```

O valor de retorno de `repeat` é `None`.

Agora aqui está uma função parecida com `repeat`, exceto que ela tem um valor de retorno.

```python
def repeat_string(word, n):
    return word * n
```

Note que podemos usar uma expressão em um comando `return`, não apenas uma variável.

Com essa versão, podemos atribuir o resultado a uma variável.
Quando a função é executada, ela não exibe nada.

```python
line = repeat_string('Hello, ', 4)
```

Mas depois podemos exibir o valor atribuído a `line`.

```python
line
```

Uma função como essa é chamada de **função pura**, porque não exibe nada nem tem qualquer outro efeito, além de retornar um valor.

## Valores de retorno e condicionais

Se o Python não fornecesse `abs`, poderíamos escrevê-la assim.

```python
def absolute_value(x):
    if x < 0:
        return -x
    else:
        return x
```

Se `x` for negativo, o primeiro comando `return` retorna `-x` e a função termina imediatamente.
Caso contrário, o segundo comando `return` retorna `x` e a função termina.
Então essa função está correta.

No entanto, se você colocar comandos `return` dentro de um condicional, precisa se certificar de que todo caminho possível pelo programa chegue a um comando `return`.
Por exemplo, aqui está uma versão incorreta de `absolute_value`.

```python
def absolute_value_wrong(x):
    if x < 0:
        return -x
    if x > 0:
        return x
```

Aqui está o que acontece se chamarmos essa função com `0` como argumento.

```python
absolute_value_wrong(0)
```

Não obtemos nada! Aqui está o problema: quando `x` é `0`, nenhuma das condições é verdadeira, e a função termina sem chegar a um comando `return`, o que significa que o valor de retorno é `None`, então o Jupyter não exibe nada.

Como outro exemplo, aqui está uma versão de `absolute_value` com um comando `return` extra no final.

```python
def absolute_value_extra_return(x):
    if x < 0:
        return -x
    else:
        return x

    return 'Código morto'
```

Se `x` for negativo, o primeiro comando `return` é executado e a função termina. Caso contrário, o segundo comando `return` é executado e a função termina.
De qualquer forma, nunca chegamos ao terceiro comando `return`, ou seja, ele nunca será executado.

Código que nunca é executado é chamado de **código morto** (*dead code*). Em geral, código morto não causa nenhum dano, mas geralmente indica um mal-entendido, e pode confundir alguém que está tentando entender o programa.

## Desenvolvimento incremental

Conforme você escreve funções maiores, pode se ver gastando mais
tempo depurando. Para lidar com programas cada vez mais complexos, você pode
querer experimentar o **desenvolvimento incremental**, que é uma forma de
adicionar e testar apenas uma pequena quantidade de código por vez.

Como exemplo, suponha que você queira encontrar a distância entre dois pontos representados pelas coordenadas $(x_1, y_1)$ e $(x_2, y_2)$.
Pelo teorema de Pitágoras, a distância é:

$$\mathrm{distance} = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$

O primeiro passo é considerar como uma função `distance` deveria se parecer em Python, ou seja, quais são as entradas (parâmetros) e qual é a saída (valor de retorno)?

Para essa função, as entradas são as coordenadas dos pontos.
O valor de retorno é a distância.

```python
# x1 = 1
# y1 = 2
# x2 = 4
# y2 = 6
```

```python
def distance(x1, y1, x2, y2):
  return 0.0
```

Essa versão ainda **não calcula distâncias**, sempre retorna zero. Mas é uma **função completa**, com um valor de retorno, o que significa que você pode testá-la antes de torná-la mais complexa.

Para testar a nova função, vamos chamá-la com argumentos de exemplo:

```python
distance(1, 2, 4, 6)
```

Neste ponto **confirmamos** que a função é **executada** e **retorna** um valor, e podemos começar a adicionar código ao corpo dela.
Um bom próximo passo é encontrar as diferenças `x2 - x1` e `y2 - y1`.
Aqui está uma versão que armazena esses valores em variáveis temporárias e os exibe.

```python
def distance(x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1
  print(f"dx = {dx}")
  print(f"dy = {dy}")
  return 0.0
```

Se a função estiver funcionando, ela deve exibir `dx é 3` e `dy é 4`.
Se isso acontecer, sabemos que a função está recebendo os argumentos
corretos e realizando o primeiro cálculo corretamente. Se não, há apenas
algumas linhas para verificar.

```python
distance(1, 2, 4, 6)
```

Em seguida, calculamos a soma dos quadrados de `dx` e `dy`:

```python
def distance(x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1
  dsquare = (dx ** 2) + (dy ** 2)
  print(f"dsquare = {dsquare}")
  return 0.0
```

Novamente, podemos rodar a função e verificar a saída, que deve ser `25`.

```python
distance(1, 2, 4, 6)
```

Por fim, podemos usar `math.sqrt` para calcular a distância:

```python
def distance(x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1
  dsquare = (dx ** 2) + (dy ** 2)
  resultado = math.sqrt(dsquare)
  print(f"resultado = {resultado}")
  return 0.0
```

E testá-la.

```python
distance(1, 2, 4, 6)
```

O resultado está correto, mas essa versão da função exibe o resultado em vez de retorná-lo, então o valor de retorno é `None`.

Podemos corrigir isso substituindo a função `print` por um comando `return`.

```python
def distance(x1, y1, x2, y2):
  dx = x2 - x1
  dy = y2 - y1
  dsquare = (dx ** 2) + (dy ** 2)
  return math.sqrt(dsquare)
```

Essa versão de `distance` é uma função pura.
Se a chamarmos assim, apenas o resultado é exibido.

```python
resultado = distance(1, 2, 4, 6)
```

E se atribuirmos o resultado a uma variável, nada é exibido.

```python
print(f"A distância é {resultado}")
```

Os comandos `print` que escrevemos são úteis para depuração, mas, uma vez que a função esteja funcionando, podemos removê-los.

Esse exemplo demonstra o desenvolvimento incremental.
Os aspectos principais desse processo são:

1.  Comece com um programa que funciona, faça pequenas mudanças e teste após cada mudança.

2.  Use variáveis para guardar valores intermediários, para que você possa exibi-los e conferi-los.

3.  Depois que o programa estiver funcionando, remova as variáveis intermediárias.

A qualquer momento, se houver um erro, você deve ter uma boa ideia de onde ele está. O desenvolvimento incremental pode economizar bastante tempo de depuração.

## Funções booleanas

Funções podem retornar os valores booleanos `True` e `False`, o que costuma ser conveniente para encapsular um teste complexo em uma função.
Por exemplo, `is_divisible` verifica se `x` é divisível por `y` sem resto.

```python
def is_divisible(x, y):
    if x % y == 0:
        return True
    else:
        return False
```

Veja como usamos ela.

```python
is_divisible(6, 4)
```

```python
is_divisible(6, 3)
```

Dentro da função, o resultado do operador `==` é um booleano, então podemos escrever a
função de forma mais concisa, retornando-o diretamente.

```python
def is_divisible(x, y):
    return x % y == 0
```

Funções booleanas são frequentemente usadas em comandos condicionais.

```python
if is_divisible(6, 2):
    print('divisible')
```

Também podemos escrever algo assim:

```python
if is_divisible(6, 2) == True:
    print('divisible')
```

Mas a **comparação é desnecessária**.

## Glossário

**valor de retorno:**
O resultado de uma função. Se a chamada de uma função for usada como expressão, o valor de retorno é o valor da expressão.

**função pura:**
Uma função que não exibe nada nem tem qualquer outro efeito, além de retornar um valor de retorno.


**código morto:**
Parte de um programa que nunca pode ser executada, geralmente porque aparece depois de um comando `return`.

**desenvolvimento incremental:**
Um plano de desenvolvimento de programas que busca evitar depuração adicionando e testando apenas uma pequena quantidade de código por vez.
