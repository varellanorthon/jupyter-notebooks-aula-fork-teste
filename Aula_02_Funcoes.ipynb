# Funções

Até este ponto, usamos várias **funções fornecidas pelo Python**, como `int` e `float`, e algumas fornecidas pelo módulo `math`, como `sqrt` e `pow`. Agora, você aprenderá como criar suas próprias funções e executá-las.

A capacidade de escrever suas próprias funções é a base da programação.

## Definindo novas funções

Uma **definição de função** especifica o nome de uma nova função e a sequência de comandos que são executados quando a função é chamada. Aqui está um exemplo:

```python
def print_bug():
  '''
  FunçãO que imprime dois trechos de um bug.
  '''
  print("Funciona na minha máquina..")
  print("Deve ser erro no ambiente de produção!")
```

`def` é uma **palavra-chave** que indica que isso é uma definição de função.
O nome da função é `print_bug`. Qualquer coisa que seja um nome de variável válido também é um nome de função válido.

Os parênteses vazios após o nome indicam que essa função **não recebe nenhum argumento**.

A primeira linha da definição da função é chamada de **cabeçalho** e o restante é chamado de **corpo**.
O cabeçalho tem que terminar com **dois-pontos** e o corpo tem que ser **indentado**. Por convenção, a indentação é sempre de **quatro espaços**.
O corpo dessa função é composto por dois comandos print; em geral, o corpo de uma função pode conter qualquer número de comandos de qualquer tipo.

Definir uma função cria um **objeto função**, que podemos exibir assim.

```python
print_bug
```

A saída indica que `print_bug` é uma função que não recebe argumentos.
`__main__` é o nome do módulo que contém `print_bug`.

Agora que definimos uma função, podemos chamá-la da mesma forma que chamamos funções embutidas.

```python
print_bug()
```

Quando a função é executada, ela executa os comandos do corpo, que exibem as duas linhas da função `print`.

## Parâmetros

Algumas das funções **exigem argumentos**; por exemplo, quando você chama `abs` você passa um número como argumento. Algumas funções recebem mais de um argumento; por exemplo, `math.pow` recebe dois, a base e o expoente.

Aqui está a definição de uma função que recebe um argumento.

```python
def print_twice(trecho):
    print(trecho)
    print(trecho)
```

O nome da variável entre parênteses é um **parâmetro**.
Quando a função é chamada, **o valor do argumento é atribuído ao parâmetro**. Por exemplo, podemos chamar `print_twice` assim.

```python
print_twice('Hello World, ')
```

Executar essa função tem o mesmo efeito que atribuir o argumento ao parâmetro e depois executar o corpo da função, assim.

```python
string = 'Hello World, '
print(string)
print(string)
```

Você também pode usar uma **variável como argumento**.

```python
line = 'Hello World, '
print_twice(line)
```

Neste exemplo, o valor de `line` é atribuído ao parâmetro `string`.

## Chamando funções

Uma vez que você definiu uma função, você pode usá-la **dentro de outra função**. Para demonstrar, vamos escrever funções que exibem a letra de "The Spam Song" (<https://www.songfacts.com/lyrics/monty-python/the-spam-song>).

> Spam, Spam, Spam, Spam,  
> Spam, Spam, Spam, Spam,  
> Spam, Spam,  
> (Lovely Spam, Wonderful Spam!)  
> Spam, Spam,

Vamos começar com a seguinte função, que recebe dois parâmetros.


```python
def repeat(palavra, n):
    print(palavra * n)
```

Podemos usar essa função para exibir a primeira linha da música, assim.

```python
spam = 'Spam, '
repeat(spam, 4)
```

Para exibir as duas primeiras linhas, podemos definir uma nova função que usa `repeat`.

```python
def first_two_lines():
    repeat(spam, 4)
    repeat(spam, 4)
```

E então chamá-la assim.

```python
first_two_lines()
```

Para exibir as três últimas linhas, podemos definir outra função, que também usa `repeat`.

```python
def last_three_lines():
    repeat(spam, 2)
    print('(Lovely Spam, Wonderful Spam!)')
    repeat(spam, 2)
```

```python
last_three_lines()
```

Por fim, podemos juntar tudo em uma função que exibe a estrofe inteira.

```python
def print_verse():
    first_two_lines()
    last_three_lines()
```

```python
print_verse()
```

Quando executamos `print_verse`, ela chama `first_two_lines`, que chama `repeat`, que chama `print`.

## Repetição

Se quisermos exibir mais de uma estrofe, podemos usar um comando `for`.

```python
for i in range(10):
    print(i)
```

A primeira linha é um cabeçalho que termina com dois-pontos.
A segunda linha é o corpo, que precisa ser indentado.

O **cabeçalho** começa com a palavra-chave `for`, uma nova variável chamada `i`, e outra palavra-chave, `in`.
Ele usa a função `range` para criar uma sequência de dois valores, que são `0` e `1`.
Em Python, quando começamos a contar, geralmente começamos do `0`.

Quando o comando `for` é executado, ele atribui o primeiro valor de `range` a `i` e então executa a função `print` no corpo, que exibe `0`.

Quando chega ao final do corpo, ele volta para o cabeçalho, por isso esse comando é chamado de **laço**.
Na segunda vez pelo laço, ele atribui o próximo valor de `range` a `i`, e o exibe.
Então, como esse é o último valor de `range`, o laço termina.

Aqui está como podemos usar um laço `for` para exibir duas estrofes da música.

```python
for i in range(2):
    print("Verso", i)
    print_verse()
    print()
```

Você pode colocar um laço `for` dentro de uma função.
Por exemplo, `print_n_verses` recebe um parâmetro chamado `n`, que tem que ser um número inteiro, e exibe o número informado de estrofes.

```python
def print_n_verses(n):
    for i in range(n):
        print_verse()
        print()
```

```python
print_n_verses(3)
```

Neste exemplo, não usamos `i` no corpo do laço, mas ainda assim precisa haver um nome de variável no cabeçalho.

## Variáveis e parâmetros são locais

Quando você cria uma variável dentro de uma função, ela é **local**, o que
significa que ela só existe dentro da função.
Por exemplo, a função a seguir recebe dois argumentos, os concatena e exibe o resultado duas vezes.

```python
def imprime(part1, part2):
    palavras = part1 + part2
    print_twice(palavras)
```

```python
palavras = 'teste'
```

```python
palavras
```

Aqui está um exemplo que a utiliza:

```python
line1 = 'Always look on the '
line2 = 'bright side of life.'
imprime(line1, line2)
```

Quando `imprime` é executada, ela cria uma variável local chamada `palavras`, que é destruída quando a função termina.
Se tentarmos exibi-la, obtemos um `NameError`:

```python
print(palavras)
```

Fora da função, `palavras` não está definida.

Parâmetros também são locais. Por exemplo, fora de `imprime`, não existe algo como `part1` ou `part2`.

## Por que usar funções?

Talvez ainda não esteja claro por que vale a pena o trabalho de dividir um programa em funções. Há várias razões:

-   Criar uma nova função dá a você a oportunidade de **nomear um grupo de comandos**, o que torna seu programa mais fácil de ler e depurar.

-   Funções podem tornar um programa menor ao **eliminar código repetitivo**. Mais tarde, se você fizer uma alteração, só precisará fazê-la em um lugar.

-   Dividir um programa longo em funções permite **depurar as partes uma de cada vez** e depois montá-las em um todo funcional.

-   Funções bem projetadas costumam ser úteis para muitos programas. Uma vez que você as escreve e depura, **pode reutilizá-las**.

## Glossário

**definição de função:**
Um comando que cria uma função.

**cabeçalho:**
 A primeira linha de uma definição de função.

**corpo:**
 A sequência de comandos dentro de uma definição de função.

**objeto função:**
Um valor criado por uma definição de função.
O nome da função é uma variável que se refere a um objeto função.

**parâmetro:**
 Um nome usado dentro de uma função para se referir ao valor passado como argumento.

**laço:**
 Um comando que executa um ou mais comandos, geralmente de forma repetida.

**variável local:**
Uma variável definida dentro de uma função, e que só pode ser acessada dentro da função.

**traceback:**
 Uma lista das funções que estão em execução, impressa quando ocorre uma exceção.
