## A instrução `import`

Para usar alguns recursos do Python, é preciso **importá-los**.
Por exemplo, a instrução a seguir importa o módulo `math`.

```python
import math
```

Um **módulo** é uma coleção de variáveis e funções.
O módulo math fornece uma variável chamada `pi` que contém o valor da constante matemática denotada por $\pi$.
Podemos exibir seu valor assim.

```python
math.pi
```

```python
from math import pi

pi
```

Para usar uma variável de um módulo, é preciso usar o **operador ponto** (`.`) entre o nome do módulo e o nome da variável.

O módulo math também contém funções.
Por exemplo, `sqrt` calcula raízes quadradas.

```python
math.sqrt(25)
```

E `pow` eleva um número à potência de um segundo número.

```python
math.pow(5, 2)
```

Neste ponto, já vimos duas formas de elevar um número a uma potência: podemos usar a função `math.pow` ou o operador de exponenciação, `**`.
Qualquer uma das duas funciona bem, mas o operador é usado com mais frequência do que a função.

## A função print

Quando você avalia uma expressão, o resultado é exibido.

```python
n = 19
n + 1
```

Mas se você avaliar mais de uma expressão, apenas o valor da última é exibido.

```python
n + 2
n + 3
```

Para exibir mais de um valor, você pode usar a função `print`.

```python
print(n+2)
print(n+3)
```

Ela também funciona com números de ponto flutuante e strings.

```python
print('O valor de pi é aproximadamente')
print(math.pi)
```

Você também pode usar uma sequência de expressões separadas por vírgulas.

```python
print('O valor de pi é aproximadamente', math.pi)
```

Observe que a função `print` coloca um espaço entre os valores.

## Argumentos

Quando você chama uma função, a expressão entre parênteses é chamada de **argumento**.

Algumas das funções que vimos até agora recebem apenas um argumento, como `int`.

```python
int('101')
```

Algumas recebem dois, como `math.pow`.

```python
math.pow(5, 2)
```

Algumas podem receber argumentos adicionais que são opcionais.
Por exemplo, `int` pode receber um segundo argumento que especifica a base do número.

```python
int('101', 2)
```

A sequência de dígitos `101` na base 2 representa o número 5 na base 10.

`round` também recebe um segundo argumento opcional, que é o número de casas decimais para arredondar.

```python
round(math.pi, 10)
```

Algumas funções podem receber qualquer número de argumentos, como `print`.

```python
print('Qualquer', 'número', 'de', 'argumentos')
```

Se você chamar uma função e fornecer argumentos demais, isso gera um `TypeError`.

```python
%%expect TypeError

float('123.0', 2)
```

Se você fornecer poucos argumentos, isso também gera um `TypeError`.

```python
math.pow(2)
```

E se você fornecer um argumento com um tipo que a função não consegue lidar, isso também gera um `TypeError`.

```python
math.sqrt('123')
```

## Comentários

À medida que os programas ficam maiores e mais complicados, eles ficam mais difíceis de ler.
Linguagens formais são densas, e muitas vezes é difícil olhar para um trecho de código e entender o que ele está fazendo e por quê.

Por esse motivo, é uma boa ideia adicionar notas aos seus programas para explicar, em linguagem natural, o que o programa está fazendo.
Essas notas são chamadas de **comentários**, e começam com o símbolo `#`.

```python
# número de segundos em 42:42
seconds = 42 * 60 + 42
```

Neste caso, o comentário aparece sozinho em uma linha. Você também pode colocar
comentários no final de uma linha:

```python
miles = 10 / 1.61     # 10 km em milhas
```

Tudo, desde o `#` até o final da linha, é ignorado - não tem
nenhum efeito na execução do programa.

Comentários são mais úteis quando documentam características não óbvias do código.
É razoável supor que o leitor consiga descobrir *o que* o código faz; é mais útil explicar *por quê*.

Este comentário é **redundante** com o código e inútil:

```python
v = 8     # atribui 8 para a variável v
```

Este comentário contém informação útil que não está no código:

```python
v = 8     # velocidade em milhas por hora
```

Bons nomes de variáveis podem reduzir a necessidade de comentários, mas nomes longos podem tornar expressões complexas difíceis de ler, então há um compromisso (*tradeoff*) entre os dois.

## Depuração (Debugging)

Três tipos de erros podem ocorrer em um programa: **erros de sintaxe**, **erros de execução** (*runtime*) e **erros semânticos**.
É útil distingui-los para conseguir identificá-los mais rapidamente.

* **Erro de sintaxe**: "Sintaxe" se refere à estrutura de um programa e às regras sobre essa estrutura. Se houver um erro de sintaxe em qualquer parte do seu programa, o Python não executa o programa. Ele exibe uma mensagem de erro imediatamente.

* **Erro de execução (*runtime error*)**: Se não houver erros de sintaxe no seu programa, ele pode começar a rodar. Mas se algo der errado, o Python exibe uma mensagem de erro e para. Esse tipo de erro é chamado de erro de execução.

* **Erro semântico**: O terceiro tipo de erro é o "semântico". Se houver um erro semântico no seu programa, ele roda sem gerar mensagens de erro, mas não faz o que você pretendia. Identificar erros semânticos pode ser complicado, pois exige que você trabalhe de trás para frente, observando a saída do programa e tentando descobrir o que ele está fazendo.

Um nome de **variável inválido** é um erro de **sintaxe**.

```python
milhao! = 1000000
```

Se você usar um **operador com um tipo que ele não suporta**, isso é um erro de **execução**.

```python
'126' / 3
```

Por fim, aqui está um exemplo de erro **semântico**.
Suponha que queremos calcular a média entre `1` e `3`, mas esquecemos a ordem das operações e escrevemos isto:

```python
1 + 3 / 2
```

Quando essa expressão é avaliada, ela não produz uma mensagem de erro, então não há erro de sintaxe nem erro de execução.
Mas o resultado não é a média entre `1` e `3`, então o programa não está correto.
Isso é um erro semântico porque o programa executa, mas não faz o que era pretendido.

## Glossário

**instrução import (import statement):**
Uma instrução que lê um arquivo de módulo para que possamos usar as variáveis e funções que ele contém.

**módulo (module):**
Um arquivo que contém código Python, incluindo definições de função e, às vezes, outras instruções.

**operador ponto (dot operator):**
O operador `.`, usado para acessar uma função de outro módulo especificando o nome do módulo seguido de um ponto e o nome da função.

**avaliar (evaluate):**
Realizar as operações de uma expressão para calcular um valor.

**instrução (statement):**
Uma ou mais linhas de código que representam um comando ou ação.

**executar (execute):**
Rodar uma instrução e fazer o que ela diz.

**argumento (argument):**
Um valor fornecido a uma função quando ela é chamada.

**comentário (comment):**
Texto incluído em um programa que fornece informações sobre o programa, mas não tem efeito na sua execução.

**erro de execução (runtime error):**
Um erro que faz um programa exibir uma mensagem de erro e ser encerrado.

**exceção (exception):**
Um erro que é detectado enquanto o programa está em execução.

**erro semântico (semantic error):**
Um erro que faz um programa fazer a coisa errada, mas sem exibir uma mensagem de erro.
